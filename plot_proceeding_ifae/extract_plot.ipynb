```python
"""
extract_plot_from_pdf.py
 
Estrae le curve (istogrammi/step-plot vettoriali) da un PDF generato da
matplotlib/ROOT ecc. e ricostruisce il plot in Python.
 
Funziona SOLO se il PDF e' vettoriale (non un'immagine raster incapsulata).
Verifica: page.get_drawings() deve restituire oggetti con centinaia di
segmenti 'l' (line), non poche forme.
 
Richiede:
    pip install pymupdf matplotlib numpy --break-system-packages
 
Uso:
    1. Modifica la sezione CONFIG qui sotto (percorso PDF, etichette degli
       assi cosi' come appaiono nel PDF, indici delle curve da estrarre).
    2. Esegui: python3 extract_plot_from_pdf.py
    3. Lo script stampa gli indici dei "drawings" trovati (utile la prima
       volta per capire quali indici corrispondono alle curve da estrarre),
       poi salva JSON/CSV con i dati e un PNG di verifica.
"""


def alpha_blend(color, alpha=0.3, background='white'):
    """
    Calcola il colore risultante da:
        color con alpha
    sopra:
        background

    Il risultato è un colore solido, quindi compatibile con EPS.
    """
    fg = np.array(mcolors.to_rgb(color))
    bg = np.array(mcolors.to_rgb(background))

    rgb = alpha * fg + (1 - alpha) * bg

    return rgb

 
# ============================== CONFIG ===================================

CURVE_LABELS = [r'$\mu$ stopping', r'$\mu$ MIP', r'$p$ stopping', r'$p$ interacting', r'$\pi$ stopping', r'$\pi$ interacting']

#CURVE_LABELS = ['FV', 'barycenter + PMT/CRT', 'containment', 'muon ID', 'proton ID', 'no pion', r'$1\mu Np$', 'muon ID (new PID)', 'proton ID (new PID)', 'no pion (new PID)', r'$1\mu Np$ (new PID)']

#CURVE_COLORS = [
#    'cornflowerblue', 'deepskyblue', 'darkgreen',
#    'black', 'orange', 'red',
#]

#CURVE_COLORS = ['red', 'orange', 'brown', 'blue', 'violet', 'pink', 'black', 'yellow', 'gray', 'green', 'cyan']

CURVE_COLORS = ["#cc4d00",
            "#e69900",
            "#009980",
            "#a87fff",
            "black",
            "#1f83e7"]

# ===========================================================================
 
 
def calibrate_axes(page, x_labels, y_labels):
    """Ricava la trasformazione pixel->dati leggendo le etichette dei tick."""
    text = page.get_text('dict')
    xticks, yticks = {}, {}

    print("\n--- TESTO LETTO DAL PDF ---")

    for block in text['blocks']:
        if 'lines' not in block:
            continue

    for line in block['lines']:
        for span in line['spans']:
            t = span['text'].strip()

            if '-' in t or '−' in t or any(c.isdigit() for c in t):
                print(
                    repr(t),
                    [hex(ord(c)) for c in t]
                )

    print("--- FINE TESTO ---\n")
 
    for block in text['blocks']:
        if 'lines' not in block:
            continue
        for line in block['lines']:
            for span in line['spans']:
                t = span['text'].strip()
                x0, y0, x1, y1 = span['bbox']
                cx, cy = (x0 + x1) / 2, (y0 + y1) / 2
                if t in x_labels and t not in xticks:
                    xticks[t] = cx
                if t in y_labels and t not in yticks:
                    yticks[t] = cy
 
    missing_x = set(x_labels) - set(xticks)
    missing_y = set(y_labels) - set(yticks)
    if missing_x or missing_y:
        raise ValueError(
            f"Etichette non trovate nel PDF: X={missing_x} Y={missing_y}\n"
            "Controlla che X_TICK_LABELS/Y_TICK_LABELS corrispondano "
            "esattamente al testo presente nel PDF."
        )
 
    xs = np.array([float(k) for k in xticks])
    xpix = np.array([xticks[k] for k in xticks])
    ys = np.array([float(k) for k in yticks])
    ypix = np.array([yticks[k] for k in yticks])
 
    mx, bx = np.polyfit(xpix, xs, 1)
    my, by = np.polyfit(ypix, ys, 1)
 
    r_x = np.corrcoef(xpix, xs)[0, 1]
    r_y = np.corrcoef(ypix, ys)[0, 1]
    print(f"Calibrazione asse X: R={r_x:.6f}")
    print(f"Calibrazione asse Y: R={r_y:.6f}")
    if min(abs(r_x), abs(r_y)) < 0.999:
        print("ATTENZIONE: calibrazione poco lineare, controlla i risultati.")
 
    px2x = lambda px: mx * px + bx
    py2y = lambda py: my * py + by
    return px2x, py2y
 
 
def list_drawings(page):
    """Stampa un riepilogo dei drawings per aiutare a identificare gli indici."""
    drawings = page.get_drawings()
    print(f"\nTrovati {len(drawings)} oggetti vettoriali nel PDF:")
    for i, d in enumerate(drawings):
        kind = d.get('type')
        n = len(d.get('items', []))
        stroke = d.get('color')
        fill = d.get('fill')
        print(f"  [{i}] tipo={kind:3s}  n_segmenti={n:4d}  "
              f"stroke={stroke}  fill={fill}")
    return drawings
 
 
def extract_curve(drawing, px2x, py2y):
    """Converte i segmenti 'l' (line) di un drawing in punti (x_dato, y_dato)."""
    pts = []
    for item in drawing['items']:
        if item[0] == 'l':
            p0, p1 = item[1], item[2]
            pts.append((p0.x, p0.y))
            pts.append((p1.x, p1.y))
 
    # rimuove duplicati consecutivi mantenendo l'ordine
    dedup = [pts[0]]
    for p in pts[1:]:
        if p != dedup[-1]:
            dedup.append(p)
 
    return [(px2x(px), py2y(py)) for px, py in dedup]
 
 
def vertices_to_bins(points):
    """Da una spezzata a gradini (step-plot) ricava (bin_left, bin_right, height)."""
    xs = [p[0] for p in points]
    ys = [p[1] for p in points]
    rows = []
    for i in range(len(xs) - 1):
        # segmento orizzontale (stessa y, x diversa) = altezza di un bin
        if abs(ys[i] - ys[i + 1]) < 1e-9 and abs(xs[i] - xs[i + 1]) > 1e-6:
            rows.append((xs[i], xs[i + 1], ys[i]))
    return rows
 
 
def main():
    doc = fitz.open(PDF_PATH)
    page = doc[0]
 
    drawings = list_drawings(page)
 
    if CURVE_INDICES is None:
        print("\nImposta CURVE_INDICES con gli indici delle curve da estrarre "
              "(guarda l'elenco sopra) e rilancia lo script.")
        return
 
    px2x, py2y = calibrate_axes(page, X_TICK_LABELS, Y_TICK_LABELS)
 
    results = {}
    for idx, label in zip(CURVE_INDICES, CURVE_LABELS):
        points = extract_curve(drawings[idx], px2x, py2y)
        results[label] = points
        print(f"{label}: {len(points)} punti estratti")
 
#    # --- salva dati grezzi (vertici) in JSON ---
#    json_path = f"{OUTPUT_DIR}/{OUTPUT_BASENAME}_extracted.json"
#    with open(json_path, 'w') as f:
#        json.dump(results, f, indent=1)
#    print(f"\nSalvato: {json_path}")
# 
#    # --- salva un CSV per curva, in formato bin_left/bin_right/height ---
#    for label in CURVE_LABELS:
#        rows = vertices_to_bins(results[label])
#        safe_label = label.replace(' ', '_')
#        csv_path = f"{OUTPUT_DIR}/{OUTPUT_BASENAME}_{safe_label}.csv"
#        with open(csv_path, 'w', newline='') as f:
#            w = csv.writer(f)
#            w.writerow(['bin_left', 'bin_right', 'height'])
#            w.writerows(rows)
#        print(f"Salvato: {csv_path} ({len(rows)} bin)")
 
    # --- ricostruisce il plot per verifica visiva ---
    plt.figure(figsize=(7, 5))
    for label, color in zip(CURVE_LABELS, CURVE_COLORS):
        pts = results[label]
        xs = [p[0] for p in pts]
        ys = [p[1] for p in pts]
        plt.plot(xs, ys, label=label, color=color, lw=2.5)

        fill_color = alpha_blend(color, 0.25)

        if True :
            plt.fill_between(
            xs,
            ys,
            0,
            step='pre',
            alpha=0.25,
            color=fill_color
            )   

 
    plt.xlabel(X_AXIS_NAME)
    plt.ylabel(Y_AXIS_NAME)
    plt.title(PLOT_TITLE, color='#e69900', fontsize=12, loc='left', fontweight='bold')
    plt.legend(fontsize=12)
 
    plot_path = f"{OUTPUT_DIR}/{OUTPUT_BASENAME}.eps"
    plt.savefig(plot_path, format='eps', bbox_inches='tight')
    #plot_path = f"{OUTPUT_DIR}/{OUTPUT_BASENAME}.pdf"
    #plt.savefig(plot_path, format='pdf', bbox_inches='tight')
    print(f"Salvato: {plot_path}")

if __name__ == "__main__":
    main()
```

```python


# Etichette nell'ordine della legenda originale
#CUSTOM_LABELS = [
#    r'$\mu$ stopping',
#    r'$\mu$ MIP',
#    r'$p$ stopping',
#    r'$p$ interacting',
#    r'$\pi$ stopping',
#    r'$\pi$ interacting',
#]

#CUSTOM_COLORS = [
#    "#cc4d00",
#    "#e69900",
#    "#009980",
#    "#a87fff",
#    "black",
#    "#1f83e7",
#]

CUSTOM_LABELS = ['before stitching','after stitching']
CUSTOM_COLORS = ['cornflowerblue','green']

# Numero minimo di segmenti per considerare un oggetto come curva
MIN_CURVE_ITEMS = 20

NUM_RE = re.compile(r'^-?\d+\.?\d*$')

# ===========================================================================


def alpha_blend(color, alpha=0.3, background='white'):
    """
    Colore risultante da 'color' con alpha sopra 'background'.
    Serve solo per il fill_between.
    """

    fg = np.array(mcolors.to_rgb(color))
    bg = np.array(mcolors.to_rgb(background))

    return alpha * fg + (1 - alpha) * bg


# --------------------------------------------------------------------------
# GEOMETRIA
# --------------------------------------------------------------------------

def bbox_of_drawing(d):
    """
    Restituisce il bounding box del drawing PDF.
    """

    xs = []
    ys = []

    for it in d['items']:

        for p in it[1:]:

            if hasattr(p, 'x'):
                xs.append(p.x)
                ys.append(p.y)

            elif hasattr(p, 'x0'):
                xs += [p.x0, p.x1]
                ys += [p.y0, p.y1]

            elif hasattr(p, 'rect'):
                r = p.rect
                xs += [r.x0, r.x1]
                ys += [r.y0, r.y1]

    if not xs:
        return None

    return min(xs), min(ys), max(xs), max(ys)


def get_color(d):
    """
    Per i segmenti 's' il colore è d['color'].
    """

    return d.get('color')


# --------------------------------------------------------------------------
# 1) RIQUADRO DEGLI ASSI
# --------------------------------------------------------------------------

def find_axes_bbox(page, drawings):

    page_area = page.rect.width * page.rect.height

    cands = []

    for d in drawings:

        if d['type'] != 'f':
            continue

        fill = d.get('fill')

        if fill is None:
            continue

        if not (
            fill[0] > 0.98
            and fill[1] > 0.98
            and fill[2] > 0.98
        ):
            continue

        bb = bbox_of_drawing(d)

        if bb is None:
            continue

        area = (
            (bb[2] - bb[0])
            *
            (bb[3] - bb[1])
        )

        if area >= 0.98 * page_area:
            continue

        cands.append((area, bb))

    cands.sort(reverse=True)

    if not cands:
        raise RuntimeError(
            "Non trovo il riquadro degli assi."
        )

    return cands[0][1]


# --------------------------------------------------------------------------
# 2) TICK DEGLI ASSI
# --------------------------------------------------------------------------

def find_tick_positions(page, axes_bbox, axis, band=16):

    x0, y0, x1, y1 = axes_bbox

    text = page.get_text('dict')

    spans = []

    for block in text['blocks']:

        if 'lines' not in block:
            continue

        for line in block['lines']:

            for span in line['spans']:

                t = span['text'].strip()

                if not NUM_RE.match(t):
                    continue

                bx0, by0, bx1, by1 = span['bbox']

                cx = (bx0 + bx1) / 2
                cy = (by0 + by1) / 2

                # ---------------- X ----------------

                if axis == 'x':

                    if (
                        by0 >= y1 - 1
                        and
                        x0 - 10 <= cx <= x1 + 10
                    ):

                        spans.append(
                            (by0, cx, t)
                        )

                # ---------------- Y ----------------

                else:

                    if (
                        bx1 <= x0 + 1
                        and
                        y0 - 10 <= cy <= y1 + 10
                    ):

                        spans.append(
                            (bx1, cy, t)
                        )

    if not spans:
        return [], []

    ref = min(s[0] for s in spans)

    kept = [
        (pos, t)
        for off, pos, t in spans
        if abs(off - ref) <= band
    ]

    found = {}

    for pos, t in kept:

        found.setdefault(
            round(pos, 1),
            t
        )

    return (
        list(found.keys()),
        [float(v) for v in found.values()]
    )


# --------------------------------------------------------------------------
# CALIBRAZIONE
# --------------------------------------------------------------------------

def sign_correct_and_fit(pix, raw, axis):

    if len(pix) < 2:

        raise RuntimeError(
            f"Trovati troppo pochi tick sull'asse {axis}."
        )

    order = np.argsort(pix)

    pix_s = np.array(pix)[order]
    raw_s = np.array(raw)[order]

    best_k = 0
    best_r = -1

    for k in range(len(raw_s) + 1):

        vals = np.array([
            -v if i < k else v
            for i, v in enumerate(raw_s)
        ])

        if len(set(vals)) < 2:
            continue

        r = abs(
            np.corrcoef(
                pix_s,
                vals
            )[0, 1]
        )

        if r > best_r:

            best_r = r
            best_k = k

    vals = np.array([
        -v if i < best_k else v
        for i, v in enumerate(raw_s)
    ])

    m, b = np.polyfit(
        pix_s,
        vals,
        1
    )

    print(
        f"Calibrazione asse {axis.upper()}: "
        f"n_tick={len(pix_s)}  "
        f"R={best_r:.6f}"
        +
        (
            "  (corretto segno meno mancante)"
            if best_k not in (0, len(raw_s))
            else ""
        )
    )

    if best_r < 0.999:

        print(
            f"ATTENZIONE: fit poco lineare "
            f"sull'asse {axis} "
            f"(R={best_r:.4f})."
        )

    print(
        f"  pixel: {pix_s}"
    )

    print(
        f"  valori: {vals}"
    )

    print(
        f"  trasformazione: "
        f"data = {m:.8f} * pixel + {b:.8f}"
    )

    return lambda p: m * p + b


# --------------------------------------------------------------------------
# 3) TROVA LE CURVE USANDO SOLO 's'
# --------------------------------------------------------------------------

def find_curves(page, drawings, min_items=MIN_CURVE_ITEMS):

    text = page.get_text('dict')

    text_spans = []

    for block in text['blocks']:

        if 'lines' not in block:
            continue

        for line in block['lines']:

            for span in line['spans']:

                t = span['text'].strip()

                is_number = (
                    t.replace('.', '', 1)
                     .replace('-', '', 1)
                     .isdigit()
                )

                if len(t) >= 3 and not is_number:

                    x0, y0, x1, y1 = span['bbox']

                    text_spans.append(
                        (
                            x0,
                            y0,
                            x1,
                            y1,
                            t
                        )
                    )

    # ------------------------------------------------------------------
    # Trova gli elementi della legenda
    # ------------------------------------------------------------------

    swatches = []

    for d in drawings:

        # IMPORTANTE:
        # usiamo SOLO gli oggetti stroke
        if d['type'] != 'f':
            continue

        c = d.get('fill')

        if c is None:
            continue

        if len(d['items']) > 4:
            continue

        bb = bbox_of_drawing(d)

        if bb is None:
            continue

        w = bb[2] - bb[0]
        h = bb[3] - bb[1]

        if h > 30 or w > 40:
            continue

        swatches.append(
            (bb, c)
        )

    # ------------------------------------------------------------------
    # Associa colore -> label
    # ------------------------------------------------------------------

    color_to_label = {}

    for bb, color in swatches:

        cy = (bb[1] + bb[3]) / 2

        best = None
        best_dist = 1e9

        for (
            tx0,
            ty0,
            tx1,
            ty1,
            t
        ) in text_spans:

            tcy = (ty0 + ty1) / 2

            if tx0 < bb[0]:
                continue

            if abs(tcy - cy) > 8:
                continue

            dist = tx0 - bb[2]

            if 0 <= dist < best_dist:

                best_dist = dist
                best = t

        if best:

            color_to_label[
                tuple(
                    round(c, 2)
                    for c in color
                )
            ] = (
                best,
                cy
            )

    # ------------------------------------------------------------------
    # TROVA LE CURVE
    #
    # Qui è la modifica fondamentale:
    #
    # d['type'] == 's'
    #
    # e non:
    #
    # d['type'] == 'f'
    # ------------------------------------------------------------------

    hist = []

    for d in drawings:

        if d['type'] != 's':
            continue

        c = d.get('color')

        if c is None:
            continue

        if len(d['items']) < min_items:
            continue

        bb = bbox_of_drawing(d)

        if bb is None:
            continue

        hist.append(d)

    print(
        f"\nTrovati {len(hist)} oggetti "
        f"di tipo 's' candidati alle curve."
    )

    # ------------------------------------------------------------------
    # Elimina eventuali duplicati dello stesso colore
    # ------------------------------------------------------------------

    best_by_color = {}

    for d in hist:

        c = tuple(
            round(x, 2)
            for x in d['color']
        )

        if (
            c not in best_by_color
            or
            len(d['items'])
            >
            len(
                best_by_color[c]['items']
            )
        ):

            best_by_color[c] = d

    hist = list(
        best_by_color.values()
    )

    print(
        f"Dopo il raggruppamento per colore: "
        f"{len(hist)} curve."
    )

    # ------------------------------------------------------------------
    # Associa ogni curva alla legenda
    # ------------------------------------------------------------------

    results = []

    for d in hist:

        c = d['color']

        best_label = None
        best_dist = 1e9
        best_cy = None

        for lc, (
            label,
            cy
        ) in color_to_label.items():

            dist = sum(
                (a - b) ** 2
                for a, b in zip(c, lc)
            )

            if dist < best_dist:

                best_dist = dist
                best_label = label
                best_cy = cy

        results.append(
            (
                best_cy,
                best_label,
                c,
                d
            )
        )

    # legenda dall'alto verso il basso

    results.sort(
        key=lambda r: (
            r[0] is None,
            r[0]
        )
    )

    out = []

    for i, (
        cy,
        label,
        color,
        d
    ) in enumerate(results):

        if label is None:

            label = (
                f"curve_{i+1}_"
                f"rgb{tuple(round(x, 2) for x in color)}"
            )

        out.append(
            (
                label,
                color,
                d
            )
        )

    return out


# --------------------------------------------------------------------------
# 4) ESTRAZIONE DEI SEGMENTI
# --------------------------------------------------------------------------

def extract_points(drawing, px2x, py2y):

    pts = []

    for item in drawing['items']:

        if item[0] == 'l':

            p0 = item[1]
            p1 = item[2]

            pts.append(
                (
                    p0.x,
                    p0.y
                )
            )

            pts.append(
                (
                    p1.x,
                    p1.y
                )
            )

    if not pts:
        return []

    # --------------------------------------------------------------
    # Elimina punti consecutivi duplicati
    # --------------------------------------------------------------

    dedup = [pts[0]]

    for p in pts[1:]:

        if (
            abs(p[0] - dedup[-1][0]) > 1e-9
            or
            abs(p[1] - dedup[-1][1]) > 1e-9
        ):

            dedup.append(p)

    # --------------------------------------------------------------
    # Calibrazione
    # --------------------------------------------------------------

    return [
        (
            px2x(px),
            py2y(py)
        )
        for px, py in dedup
    ]


# --------------------------------------------------------------------------
# BIN
# --------------------------------------------------------------------------

def vertices_to_bins(points):

    xs = [
        p[0]
        for p in points
    ]

    ys = [
        p[1]
        for p in points
    ]

    rows = []

    for i in range(len(xs) - 1):

        if (
            abs(
                ys[i] - ys[i + 1]
            )
            < 1e-9
            and
            abs(
                xs[i] - xs[i + 1]
            )
            > 1e-6
        ):

            rows.append(
                (
                    min(
                        xs[i],
                        xs[i + 1]
                    ),

                    max(
                        xs[i],
                        xs[i + 1]
                    ),

                    ys[i]
                )
            )

    rows.sort(
        key=lambda r: r[0]
    )

    return rows


# --------------------------------------------------------------------------
# MAIN
# --------------------------------------------------------------------------

def main():

    # --------------------------------------------------------------
    # Apri PDF
    # --------------------------------------------------------------

    doc = fitz.open(
        PDF_PATH
    )

    page = doc[0]

    drawings = page.get_drawings()

    print(
        f"Dimensione pagina: {page.rect}"
    )

    print(
        f"Trovati {len(drawings)} "
        f"oggetti vettoriali nel PDF."
    )

    # --------------------------------------------------------------
    # Assi
    # --------------------------------------------------------------

    axes_bbox = find_axes_bbox(
        page,
        drawings
    )

    print(
        "\nRiquadro assi:"
    )

    print(
        f"  x_left  = {axes_bbox[0]}"
    )

    print(
        f"  y_top   = {axes_bbox[1]}"
    )

    print(
        f"  x_right = {axes_bbox[2]}"
    )

    print(
        f"  y_bottom= {axes_bbox[3]}"
    )

    # --------------------------------------------------------------
    # Tick
    # --------------------------------------------------------------

    xpix, xraw = find_tick_positions(
        page,
        axes_bbox,
        'x'
    )

    ypix, yraw = find_tick_positions(
        page,
        axes_bbox,
        'y'
    )

    print(
        "\nTick X trovati:"
    )

    print(
        list(
            zip(xpix, xraw)
        )
    )

    print("\nTick Y trovati:")

    print(list(zip(ypix, yraw)))

    # --------------------------------------------------------------
    # Calibrazione
    # --------------------------------------------------------------

    px2x = sign_correct_and_fit(xpix,xraw,'x')

    py2y = sign_correct_and_fit(ypix, yraw,'y')

    # --------------------------------------------------------------
    # Curve
    # --------------------------------------------------------------

    curves = find_curves(page, drawings)

    if not curves:
        raise RuntimeError("Nessuna curva di tipo 's' trovata automaticamente.")
    
    print("\nCurve trovate:")

    for i, (label, color, d) in enumerate(curves):

        print( f"  [{i}] label={label} color={color} segmenti={len(d['items'])}" )

    # --------------------------------------------------------------
    # Colori / labels
    # --------------------------------------------------------------

    use_custom = (len(CUSTOM_LABELS) == len(curves) and len(CUSTOM_COLORS) == len(curves))

    if not use_custom:
        print("\nATTENZIONE:")
        print(
            f"Trovate {len(curves)} curve, "
            f"ma CUSTOM_LABELS/COLORS hanno "
            f"{len(CUSTOM_LABELS)}/"
            f"{len(CUSTOM_COLORS)} elementi.")
        print("Uso le etichette originali.")

    # --------------------------------------------------------------
    # Output
    # --------------------------------------------------------------

    os.makedirs(OUTPUT_DIR, exist_ok=True)

    prepared = []

    # --------------------------------------------------------------
    # Estrai dati
    # --------------------------------------------------------------

    for i, ( auto_label, auto_color, d) in enumerate(curves):

        if use_custom:
            label = CUSTOM_LABELS[i]
            color = CUSTOM_COLORS[i]
        else:
            label = auto_label
            color = auto_color

        pts = extract_points(d, px2x, py2y)

        if not pts:
            print(f"ATTENZIONE: {label} non contiene punti.")
            continue

        xs = [ p[0] for p in pts]

        ys = [ p[1] for p in pts]

        prepared.append( (label, color, xs, ys, max(ys)) )

    # --------------------------------------------------------------
    # Disegno
    # --------------------------------------------------------------

    fig, ax = plt.subplots(figsize=(7, 5))

    for ( label, color, xs, ys, _ ) in prepared:

        ax.plot(xs, ys, label=label, color=color, lw=2.5)

        ax.fill_between( xs, ys, 0, alpha=0.25, color=alpha_blend(color, 0.25) )

    ax.set_xlabel(X_AXIS_NAME)

    ax.set_ylabel(Y_AXIS_NAME)

    ax.set_xticks(np.arange(-2,2.5,0.25))
    ax.tick_params(axis='x', rotation=60)
    ax.set_xlim(-1,1)

    ax.set_title(PLOT_TITLE, color='#e69900', fontsize=12, loc='left', fontweight='bold')

    ax.legend(fontsize=12, loc='best')

    plot_path = f"{OUTPUT_DIR}/{OUTPUT_BASENAME}.pdf"
    fig.savefig(plot_path, format='pdf', bbox_inches='tight')

if __name__ == "__main__":
    main()
```

```python
from sklearn.ensemble import GradientBoostingClassifier, AdaBoostClassifier, RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay, log_loss
from sklearn.utils.class_weight import compute_sample_weight 
from sklearn.model_selection import GridSearchCV
from skl2onnx import convert_sklearn
from skl2onnx.common.data_types import FloatTensorType
from sklearn.metrics import roc_curve, roc_auc_score

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pandas as pd

import joblib
import xml.etree.ElementTree as ET


classes_labels = ['muon_rising', 'muon mip', 'proton_rising', 'proton_interacting', 'pion_rising', 'pion_interacting']

samples = np.genfromtxt("dump_lkl_ratios_gbdt_withdepE_smallMC_newTS_newvars_only_coll.txt", names=None, dtype=float, encoding="utf-8", delimiter=" ")

samples_features = samples[:,1:samples.shape[1]]
samples_labels = samples[:,0]
X_train, X_test, y_train, y_test = train_test_split(samples_features,samples_labels,test_size=0.30,random_state=42,stratify=samples_labels)

print("The TRAINING set contains")
for i,class_name in enumerate(classes_labels):
    print(np.sum(y_train==i), '{} tracks'.format(class_name))
print("The TEST set contains")
for i,class_name in enumerate(classes_labels):
    print(np.sum(y_test==i), '{} tracks'.format(class_name))

grid_search_active = False

if grid_search_active:
    parameters = {'learning_rate':[0.025, 0.01, 0.0075, 0.005, 0.001],
              'max_depth': [3,4,5],
               'subsample': [0.75,0.5,0.4,0.3] }

    gbdt = GradientBoostingClassifier(
        loss="log_loss",     
        n_estimators=500,    
        random_state=42,
        verbose=1,
        validation_fraction=0.2, 
        n_iter_no_change=10,    
        tol=0.001               
    )

    sample_weight = compute_sample_weight(class_weight='balanced', y=y_train)
    grid_search = GridSearchCV(gbdt,parameters,cv=5,scoring='roc_auc_ovr_weighted', refit=True, n_jobs=-1)
    grid_search.fit(X_train,y_train,sample_weight=sample_weight)

    gbdt = grid_search.best_estimator_
    best_params = grid_search.best_params_
else:
    gbdt = GradientBoostingClassifier(
        loss="log_loss",     
        n_estimators=400,  
        learning_rate=0.025,
        max_depth=5,
        subsample=0.75,  
        random_state=42,
        verbose=1,
        validation_fraction=0.2, 
        n_iter_no_change=10,    
        tol=0.001               
    )
    gbdt.fit(X_train, y_train, compute_sample_weight(class_weight='balanced', y=y_train))
    best_params = gbdt.get_params()


#test
print(X_test[0], y_test[0], gbdt.predict_proba(X_test[0:1]))

######################
# test vs train loss #
######################
training_score = np.zeros(gbdt.n_estimators_)
for i,y_pred_stage in enumerate(gbdt.staged_predict_proba(X_train)):
    training_score[i] = log_loss(y_train, y_pred_stage)

test_score = np.zeros(gbdt.n_estimators_)
for i,y_pred_stage in enumerate(gbdt.staged_predict_proba(X_test)):
    test_score[i] = log_loss(y_test, y_pred_stage)

plt.plot(training_score, label='training loss', color='orange')
plt.plot(test_score, label='test loss', color='green')
plt.xlabel("Iteration number", fontsize=18)
plt.ylabel("Multinomial NLL loss", fontsize=18)
plt.legend()
plt.title("training/test deviance\nlearning rate: {} max depth: {} subsample frac: {}".format(best_params['learning_rate'],best_params['max_depth'],best_params['subsample']), fontsize=18)
plt.savefig("AAA_train_test_loss_{}_{}_{}.pdf".format(best_params['learning_rate'],best_params['max_depth'],best_params['subsample']), format="pdf", bbox_inches="tight")
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.close()

gap = np.mean(np.array(test_score[-10:]) - np.array(training_score[-10:]))
print(f"\nTrain-test loss gap (ultimi 10 step): {gap:.4f}")

#####################
# apply on test set #
#####################
y_pred = gbdt.predict(X_test)

#####################
# efficiency matrix #
#####################
disp_confusion_matrix_efficiency = ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=classes_labels, normalize='true', cmap=plt.cm.Blues)
plt.xticks(rotation=60)
plt.title("efficiency confusion matrix\nlearning rate: {} max depth: {} subsample frac: {}".format(best_params['learning_rate'],best_params['max_depth'],best_params['subsample']), fontsize=18)
plt.savefig("AAA_efficiency_confusion_matrix_{}_{}_{}.pdf".format(best_params['learning_rate'],best_params['max_depth'],best_params['subsample']), format='pdf', bbox_inches='tight')
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.xlabel('predicted label', fontsize=18)
plt.ylabel('true label', fontsize=18)
plt.close()

#################
# purity matrix #
#################
disp_confusion_matrix_purity = ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=classes_labels, normalize='pred', cmap=plt.cm.Blues)
plt.xticks(rotation=60)
plt.title("purity confusion matrix\nlearning rate: {} max depth: {} subsample frac: {}".format(best_params['learning_rate'],best_params['max_depth'],best_params['subsample']), fontsize=18)
plt.savefig("AAA_purity_confusion_matrix_{}_{}_{}.pdf".format(best_params['learning_rate'],best_params['max_depth'],best_params['subsample']), format='pdf', bbox_inches='tight')
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.xlabel('predicted label', fontsize=18)
plt.ylabel('true label', fontsize=18)
plt.close()

##############
# row matrix #
##############
disp_confusion_matrix_row = ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=classes_labels, normalize=None, cmap=plt.cm.Blues)
plt.xticks(rotation=60)
plt.title("confusion matrix\nlearning rate: {} max depth: {} subsample frac: {}".format(best_params['learning_rate'],best_params['max_depth'],best_params['subsample']), fontsize=18)
plt.savefig("AAA_row_confusion_matrix_{}_{}_{}.pdf".format(best_params['learning_rate'],best_params['max_depth'],best_params['subsample']), format='pdf', bbox_inches='tight')
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.xlabel('predicted label', fontsize=18)
plt.ylabel('true label', fontsize=18)
plt.close()
                
########################
# ROC curves 1 vs rest #
########################

n_classes = len(classes_labels)
y_test_bin = np.zeros((len(y_test), n_classes))
for j in range(n_classes):
    y_test_bin[:, j] = (y_test == j).astype(int)

y_proba = gbdt.predict_proba(X_test)
for i, name in enumerate(classes_labels):
    fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_proba[:, i])
    auc = roc_auc_score(y_test_bin[:, i], y_proba[:, i])
    plt.plot(fpr, tpr, label=f'AUC = {auc:.3f}', color='steelblue', linewidth=2)
    plt.plot([0, 1], [0, 1], 'k--', label='random classifier')
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    plt.xlabel('False Positive Rate', fontsize=18)
    plt.ylabel('True Positive Rate', fontsize=18)
    plt.title(f'ROC curve — {name}', fontsize=18)
    plt.legend()
    plt.savefig(f'AAA_roc_{name}.pdf', format= 'pdf', bbox_inches='tight')
    plt.close()

#######################
# fuature importances #
#######################

importances = gbdt.feature_importances_
feature_names = [r"$\mu$ ris / $\mu$ mip", 
                r"$\mu$ ris / p ris",
                r"$\mu$ ris / p int",
                r"$\mu$ ris / $\pi$ ris",
                r"$\mu$ ris / $\pi$ int",
                r"$\mu$ mip / p ris",
                r"$\mu$ mip / p int",
                r"$\mu$ mip / $\pi$ ris",
                r"$\mu$ mip / $\pi$ int",
                r"p ris / p int",
                r"p ris / $\pi$ ris",
                r"p ris / $\pi$ int",
                r"p int / $\pi$ ris",
                r"p int / $\pi$ int",
                r"$\pi$ ris / $\pi$ int",
                'depE',
                'daughter_depE',
                'mother/daughter angle']
plt.bar(feature_names, importances)
plt.xticks(fontsize=14, rotation=90)
plt.yticks(fontsize=14)
plt.xlabel('feature', fontsize=18)
plt.ylabel('feature importance', fontsize=18)
plt.savefig(f'AAA_feature_importance.pdf', format= 'pdf', bbox_inches='tight')
plt.close()
```
